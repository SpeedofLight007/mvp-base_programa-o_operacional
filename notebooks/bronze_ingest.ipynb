```python
import pandas as pd

BASE = "/Volumes/mvp-base_programação_operacional/bronze/raw_extraction"

df_a = pd.read_csv(f"{BASE}/sistema_a.csv", sep=";")
df_b = pd.read_excel(f"{BASE}/sistema_b.xlsx", sheet_name="data")
df_c = pd.read_excel(f"{BASE}/sistema_c_relatorio(9011).xlsx")
df_d = pd.read_excel(f"{BASE}/sistema_d.xlsx")

print(df_a.shape, df_b.shape, df_c.shape, df_d.shape)
df_a.head()
```

```python
from datetime import datetime
spark.sql("USE CATALOG `mvp-base_programação_operacional`")

data_ingestao = datetime.now().strftime('%Y-%m-%d %H:%M:%S')

df_a['_data_ingestao'] = data_ingestao
df_a['_arquivo_origem'] = 'sistema_a.csv'
df_b['_data_ingestao'] = data_ingestao
df_b['_arquivo_origem'] = 'sistema_b.xlsx'
df_c['_data_ingestao'] = data_ingestao
df_c['_arquivo_origem'] = 'sistema_c_relatorio(9011).xlsx'
df_d['_data_ingestao'] = data_ingestao
df_d['_arquivo_origem'] = 'sistema_d.xlsx'

def limpar_nomes_colunas(df):
    df.columns = [c.strip().replace(' ', '_').replace('(', '').replace(')', '') for c in df.columns]
    return df

df_a = limpar_nomes_colunas(df_a)
df_b = limpar_nomes_colunas(df_b)
df_c = limpar_nomes_colunas(df_c)
df_d = limpar_nomes_colunas(df_d)

spark.createDataFrame(df_a).write.mode('overwrite').saveAsTable('bronze.sistema_a')
spark.createDataFrame(df_b).write.mode('overwrite').saveAsTable('bronze.sistema_b')
spark.createDataFrame(df_c).write.mode('overwrite').saveAsTable('bronze.sistema_c')
spark.createDataFrame(df_d).write.mode('overwrite').saveAsTable('bronze.sistema_d')

print('4 tabelas gravadas em bronze')
```
